## Remplacons les donnees de texte par des donnees exploitables : 
La machine ne peut comprendre des chaines de caracteres lors de l'apprentissage, donc afin de lui permettre d'interpreter ces donnees, il faut les transformers en quelque chose qu'elle peut comprendre comme des donnees numeriques. (ici je le fais avec la feature LEGENDAIRE qui devient 1 si vrai ou 0 si faux)

```python
# remplacement de VRAI FAUX par 1 0
nosPokemonsClean = nosPokemons
nosPokemonsClean["LEGENDAIRE"] = nosPokemonsClean["LEGENDAIRE"].replace("VRAI", 1)
nosPokemonsClean["LEGENDAIRE"] = nosPokemonsClean["LEGENDAIRE"].replace("FAUX", 0)
nosPokemonsClean["LEGENDAIRE"] = nosPokemonsClean["LEGENDAIRE"].astype(int)
print(nosPokemonsClean.head())
```

## Extraction des donnees de `combats.csv`

## Assemblage des observations : 

### Nombre de combats : 

il est important de garder en tete qu'un pokemon ayant passe en premier, avec assez de vitesse peut donner un coup en premier, hors, s'il passe en deuxieme, il pourrait se prendre un coup en premier lors du changement de pokemon. Toutefois, cette information a une importance s'il a peu de hp car il pourrait simplement mourrir avant de donner un coup. En tout cas, il faut compter le nombre de fois ou le pokemon s'est battu en premier et l'additionner au nombre de fois ou il s'est battu en tant que second.

```python
nbFoisPremierePosition = combats.groupby('Premier_Pokemon').count()
print(nbFoisPremierePosition)
```

```python
nbFoisSecondePosition = combats.groupby('Second_Pokemon').count()
print(nbFoisSecondePosition)
```

```python
nombreTotalDeCombats = nbFoisPremierePosition + nbFoisSecondePosition
print(nombreTotalDeCombats)
```

## Interpretation importante : 

Ici, on compte le nombre de fois ou le pokemon est present en tant que premier combatant, suite a cela en tant que deuxieme, et ce que pandas fait lorsque l'on additionne ces deux variable est d'additionner les colonnes ou le nom est le meme, donc ici `Pokemon_Gagnant`, toutefois, il ne s'agit pas du nombre de fois ou le pokemon a gagnee mais plutot le nombre de combat qu'il a fait, ayant `Premier_Pokemon` comme index.

# Nombre de combats gagnes : 

Maintenant, nous passons au nombre de combats gagnes par chaque pokemon.

```python
nombreDeVictoires = combats.groupby('Pokemon_Gagnant').count()
print(nombreDeVictoires)
```

### Interpretation : 
Ici, il ne faut toujours pas prendre en compte les noms des colonnes, ici, premier pokemon et second pokemon sont juste les nombres de combats

# Agregation des donnees : 
## Creont des variables qui sont facilement interpretables : 

```python
listeAAgreger = combats.groupby('Pokemon_Gagnant').count()
listeAAgreger.sort_index()

#On ajoute le nombre de combats
listeAAgreger['NBR_COMBATS'] = nbFoisPremierePosition.Pokemon_Gagnant + nbFoisSecondePosition.Pokemon_Gagnant

#On ajoute le nombre de victoires
listeAAgreger['NBR_VICTOIRES'] = nombreDeVictoires.Premier_Pokemon

#On calcule le pourcentage de victoires
listeAAgreger['POURCENTAGE_DE_VICTOIRES'] = nombreDeVictoires.Premier_Pokemon / (nbFoisPremierePosition.Pokemon_Gagnant + nbFoisSecondePosition.Pokemon_Gagnant)

#On affiche la nouvelle liste
print(listeAAgreger)
```

Agregeons ces donnees avec le dataframe original : 

```python
nouveauPokedex = nosPokemonsClean.merge(listeAAgreger, left_on='NUMERO', right_index = True, how='left')
print(nouveauPokedex.describe())
```

## Determinons quels pokemons les dresseurs devraient prendre :

Il s'agit ici de savoir quelles probabilites de type notre adversaire aura comme pokemon afin d'avoir un type qui nous met en avantage ou en neutralite

```python
import matplotlib.pyplot as plt
import seaborn as sns

axe_X = sns.countplot(x="TYPE_1", hue="LEGENDAIRE", data=nouveauPokedex)
plt.xticks(rotation= 90)
plt.xlabel('TYPE_1')
plt.ylabel('Total ')
plt.title("POKEMONS DE TYPE_1")
plt.show()
```

## Meme chose pour les deuxiemes types : 

```python
axe_X = sns.countplot(x="TYPE_2", hue="LEGENDAIRE", data=nouveauPokedex)
plt.xticks(rotation= 90)
plt.xlabel('TYPE_2')
plt.ylabel('Total ')
plt.title("POKEMONS DE TYPE_2")
plt.show()
```

### Interpretation :

Le premier type le plus present est : “Herbe”, “Eau”, “Insecte” et “Normal”.
Le deuxieme type le plus present est : “Vol”, “Poison” et “Sol”.

## Les types qui gagnent le plus :

Connaitre les types 1 de pokemon qui gagnent le plus est a notre avantage car en les mettant dans son equipe, le dresseur aura une plus grosse tendance a gagner :

```python
print(nouveauPokedex.groupby('TYPE_1').agg({"POURCENTAGE_DE_VICTOIRES": "mean"}).sort_values(by = "POURCENTAGE_DE_VICTOIRES"))
```

### Interpretation :

les pokemons les plus gagants : Obscur • Électrique • Dragon • Vol

Les pokemons les plus perdants : Fée • Eau • Roche • Acier • Poison
